## 1. Data Cleaning and Exploratory Data Analysis (EDA)

First, we import the necessary libraries and load the dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

# Load the dataset
df = pd.read_csv('Product_Sales.csv')

# Display the first few rows of the dataframe
df.head()
```

### Handle Missing Values

```python
# Check for missing values
df.isnull().sum()
```

The output shows that there are no missing values in the dataset, so no imputation is needed.

### Convert Date Columns to Datetime

```python
# Convert 'Order Date' and 'Ship Date' to datetime objects
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d-%m-%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d-%m-%Y')

# Verify the data types
df.info()
```

The date columns have been successfully converted to datetime objects.

### Summary Statistics

```python
# Summary statistics for numerical columns
df.describe()
```

The summary statistics provide a quick overview of the numerical columns. We can see the mean, standard deviation, and quartiles for Sales, Quantity, Discount, and Profit.

### Visualizing Key Metrics

```python
# Set the style for plots
sns.set_style('whitegrid')

# Create histograms for Sales, Profit, and Discount
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.histplot(df['Sales'], bins=30, kde=True)
plt.title('Distribution of Sales')

plt.subplot(1, 3, 2)
sns.histplot(df['Profit'], bins=30, kde=True)
plt.title('Distribution of Profit')

plt.subplot(1, 3, 3)
sns.histplot(df['Discount'], bins=10, kde=True)
plt.title('Distribution of Discount')

plt.tight_layout()
plt.savefig('histograms.png')
plt.show()
```

```python
# Create boxplots for Sales, Profit, and Discount
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.boxplot(y=df['Sales'])
plt.title('Boxplot of Sales')

plt.subplot(1, 3, 2)
sns.boxplot(y=df['Profit'])
plt.title('Boxplot of Profit')

plt.subplot(1, 3, 3)
sns.boxplot(y=df['Discount'])
plt.title('Boxplot of Discount')

plt.tight_layout()
plt.savefig('boxplots.png')
plt.show()
```

The histograms and boxplots show that 'Sales' and 'Profit' are right-skewed, with some outliers. 'Discount' has a more discrete distribution.

## 2. Research Questions with Visuals

### RQ1: Which product categories are the most profitable in each region?

```python
# Group by Region and Category to find total profit
regional_profit = df.groupby(['Region', 'Category'])['Profit'].sum().reset_index()

# Create a grouped bar chart
plt.figure(figsize=(15, 8))
sns.barplot(data=regional_profit, x='Region', y='Profit', hue='Category')
plt.title('Total Profit by Product Category in Each Region')
plt.xlabel('Region')
plt.ylabel('Total Profit')
plt.legend(title='Category')
plt.tight_layout()
plt.savefig('regional_profit_by_category.png')
plt.show()
```

**Interpretation:** The bar chart shows that the 'Technology' category is the most profitable in all regions, followed by 'Office Supplies'. The 'Furniture' category is the least profitable and even incurs losses in some regions.

### RQ2: How does discount percentage impact profit?

```python
# Create a scatter plot of Discount vs. Profit
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Discount', y='Profit')
plt.title('Impact of Discount on Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.tight_layout()
plt.savefig('discount_vs_profit.png')
plt.show()
```

**Interpretation:** The scatter plot reveals a negative correlation between discount and profit. As the discount percentage increases, the profit tends to decrease. In many cases, higher discounts lead to negative profit (losses).

### RQ3: What are the sales trends over time across customer segments?

```python
# Set 'Order Date' as the index
df_time = df.set_index('Order Date')

# Resample data by month and segment
monthly_sales = df_time.groupby('Segment')['Sales'].resample('M').sum()
monthly_sales = monthly_sales.reset_index()

# Create a line plot
plt.figure(figsize=(15, 8))
sns.lineplot(data=monthly_sales, x='Order Date', y='Sales', hue='Segment')
plt.title('Monthly Sales Trends by Customer Segment')
plt.xlabel('Date')
plt.ylabel('Total Sales')
plt.legend(title='Customer Segment')
plt.tight_layout()
plt.savefig('monthly_sales_by_segment.png')
plt.show()
```

**Interpretation:** The line plot shows the sales trends for each customer segment over time. All segments follow a similar pattern, with sales peaking towards the end of the year. The 'Consumer' segment consistently has the highest sales, followed by 'Corporate' and 'Home Office'.

## 3. Advanced Data Visualizations

### Correlation Heatmap

```python
# Select numerical columns for correlation matrix
numerical_cols = df.select_dtypes(include=np.number)

# Calculate the correlation matrix
corr_matrix = numerical_cols.corr()

# Create a heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix of Numerical Variables')
plt.tight_layout()
plt.savefig('correlation_heatmap.png')
plt.show()
```

The heatmap shows the correlation between numerical variables. 'Sales' and 'Profit' have a moderate positive correlation, while 'Discount' and 'Profit' have a negative correlation.

### Interactive Scatter Plot with Plotly

```python
# Create an interactive scatter plot of Sales vs. Profit
fig = px.scatter(df, x='Sales', y='Profit', color='Category', hover_data=['Product Name'])
fig.update_layout(title='Interactive Scatter Plot of Sales vs. Profit by Category')
fig.write_html('interactive_scatter.html')
fig.show()
```

This interactive scatter plot allows for a more detailed exploration of the relationship between sales and profit, with the ability to see details for each product by hovering over the data points.
